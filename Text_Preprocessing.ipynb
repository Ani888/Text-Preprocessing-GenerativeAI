# ***remove html tag in data preprocessing***


```python
import re
def remove_html_tag(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'',text)
```

```python
text = """<html><head><title>Sentiment Analysis</title></head><body><h1>Amazing Product!</h1><p>I REALLY love this product!!! 😍</p><p>The quality is <b>excellent</b>, but the delivery was slow.</p><a href="https://example.com">Click here</a></body></html>"""
```

```python
remove_html_tag(text)
```

```python
df['review']=df['review'].apply(remove_html_tag)
```

# ***remove url in data preprocessing***

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub('', text)
```

# ***panctuation handling***

```python
import string, time
exclude = string.punctuation
exclude
```

```python
def remove_punc(text):
  for char in exclude :
    text= text.replace(char,'')
    return text
```

```python
text="""String. With. Punctuation?"""
```

```python
start=time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1*50000)
```

# ***another approach to remove punctuation***

```python
def remove_punc1(text):
  return text.translate(str.maketrans('','',exclude))
```

```python
start=time.time()
print(remove_punc1(text))
time2 = time.time() - start
print(time2*50000)
```

# ***chat conversation handle ( in this we give lots of shortcut in chat )***

```python
# create dictionary for all the shortcut first
chat_words = {
    'AFAIK':'As far as I know',
    'AFK':'Away from keyboard',
    'ASAP': 'As soon as possible'
}
```

```python
def chat_conversion(text):
  new_text=[]
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

```python
chat_conversion("Do  this  work  ASAP")
```

# ***Stop word which does not support in positive or negative sentiment***

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
  new_text =[]
  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)

  x = new_text[ : ]
  new_text.clear()
  return " ".join(x)
```

```python
remove_stopwords('This is a very good product and it is available at a reasonable price. The delivery was late, but the quality of the product is excellent. I would like to recommend this product to my friends and family. It is one of the best products in the market.')
```

# ***remove emoji handle***

```python
import re
def remove_emoji(text):
  emoji_pattern= re.compile("["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF"
        "\U0001F1E0-\U0001F1FF"
        "\U00002600-\U000026FF"
        "\U00002700-\U000027BF""]+",flags=re.UNICODE)
  return emoji_pattern.sub(r'', text)
```

```python
remove_emoji('😍😍 This product is sooo good 🔥🔥 I love the quality ❤️ but the delivery was bad 😡😡')
```

# ***Stemmer (convert 2nd and 3rd form of verb into first form. example walk, walks, walked, walking. this are all converted to walk, walk, walk, walk by stemmer)***

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_word(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample=" walk walks walking walked "
stem_word(sample)
```

```python
text=" I got very very high paying job in MNC. My role is AI/ML engineer. I will join the comapny very soon"
stem_word(text)
```

# ***Lemmatization (It change or give readable word)***

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
wordnet_lemmatizer = WordNetLemmatizer()

sentence = " I got very very high paying job in MNC. My role is AI/ML engineer. I am going to join company with two days."
punctuation ="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
  if word in punctuation:
    sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
  print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word, pos='v')))
```
